```python
# Identity Operators: is, is not

a = [1, 2, 3]
b = a          # same memory location
c = [1, 2, 3]  # different object

print(a is b)      # True
print(a is c)      # False
print(a == c)      # True (values same)
print(a is not c)  # True
```

## LOGICAL OPERATORS

           and, or, not operators

```python
x=10
y=10
print(x>5 and y<1)      #and OPERATION RETURNS True WHEN BOTH CONDITIONS ARE True. Otherwise False


print(x>5 or y<1)       #or OPERATION RETURNS True  IF ANY ONE OF THE CONDITION IS True /BOTH CONDITIONS ARE True

not(x>5 or y<1)
```

# OPERATOR PRECEDANCE

Precedence Level	              Operator(s)	                      Description

1 (Highest)	                        ()	                              Parentheses

2	                                **	                              Exponentiation

3	                                +, - (unary)	                  Unary plus, minus

4	                                *, /, //, %	                      Multiplication, Division

5	                                +, -	                          Addition, Subtraction

6	                                <, <=, >, >=	                  Comparison

7	                                ==, !=	                          Equality

8	                                not	                              Logical NOT

9	                                and	                              Logical AND

10 (Lowest)	                        or	                              Logical OR
