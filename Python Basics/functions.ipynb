```python
#write a function that checks if a number is even return the 2nd multiple of that number
def checkeven(a):
    if a%2==0:
        print(a*2)
    else:
        print("number is odd")

checkeven(11)
checkeven(20)
```

```python
#write a function that prints even numbers in an array
def printeven(numbers):
    result=[]
    for i in numbers:
      if i%2==0:
            print(i)
            result.append(i)
    print(result)        


numbers=[3,5,7,8,4,2,0]
printeven(numbers)

 
```
