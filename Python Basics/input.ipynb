## CONVERTING INPUT TO INTEGER

```python
age=int(input("Enter your age:"))   #converting input to integer by using int()
print("he is",age,"years old")
print(type(age))
```

```python
height=float(input("my height is:"))
print("your height is:",height)
print(type(height))
```

```python
num1=input("Enter first number:")
num2=input("Enter second number:")   #when we add two numbers without mention the data type(int/float/...) it will join both of them as a string
print(num1)
print(num2)
sum_result=num1+num2
print("sum=",sum_result)   #not mathamatical sum
```

```python
num1=int(input("Enter first number:"))
num2=int(input("Enter second number:"))
print(num1)
print(num2)
sum_result=num1+num2
print("sum=",sum_result)
```

## string Formatting

```python
name="Aman"
age=20
print("Name is",name,"and age is",age)
print("Hello " +name)
print("My name is {} and Iam {} years old.".format(name,age))
print("My name is {0} and Iam {1} years old.".format(name,age))
print("My name is {n} and Iam {a} years old.".format(n=name,a=age))
print(f"{name} is {age} years old.")  #fstring
```

```python
a=10
b=20
print(f"Sum = {a+b}")
print(f"Multiplication = {a*b}")
```

```python
pi=3.141592
print(f"Value of pi ={pi:.2f}")
#.2f means 2 digit after decimal
```

```python
marks=87.4567
print(f"percentage {marks:.2f}%")
```

## CONDITIONAL STATEMENT

```python
#if condition
age=20
if age>=18:
    print("you are eligible to vote")
```

```python
num=int(input("Enter a number:"))
if num > 0:
    print("positive number")
```

```python
#if_else

age=int(input("Enter your age:"))

if age>=18:
    print("Eligible for vote")
else:
    print("Not eligible for vote")
```

```python
mark=int(input("Enter your mark:"))
print("mark is:",mark)

if mark >40:
    print("your are pass")
else:
    print("you are failed")
```

```python
num=int(input("enter a number:"))

if num % 2 ==0:
    print("number is even")
else:
    print("number is odd")
```

```python
#elif
a=30
b=20
if a==b:
    print("Equal")
elif a>b:
    print("a is bigger")
else:
    print("b is bigger")
```

```python
traffic=input("Enter signal colour:")

if traffic=="red":
    print("STOP")
elif traffic=="green":
    print("GO")
else:
    print("SLOW")
```

```python
mark=int(input("Enter your mark:"))

if mark >90:
    print("A grade")
elif mark>=90:
    print("B grade")
elif mark>=80:
    print("C grade")
elif 60<mark<70:
    print("D grade")
else:
    print("Fail")
```

## NESTED IF

```python
username="admin"
password="1234"

if username=="admin":
    if password=="1234":
        print("Login successfully")
    else:
        print("Wrong password")
else:
    print("Invalid username")
```

```python
mark=int(input("Mark is:"))
income=int(input("your income:"))

if mark>=80:
    if income<=20000:
        print("eligible for scholarship")
    else:
        print("not eligible for scholarship")
else:
    print("not eligible")
```
