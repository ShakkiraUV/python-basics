#**slicing And Indexing**

```python
l=["hello","hi",23,45.6,True,12,78,False,[1,2,3,4,5,6]]
print(l[3])
```

```python
print(l[1])
```

```python
print(l[-1])
```

```python
print(l[-2])
```

```python
print(l[2:4])
```

```python
print(l[-5:-1])
```

```python
print(l[-1:-6:-1]) #when last section is -1 which mean is reverse
```

```python
print(l[::-1]) 
```

```python
print(l[-1::-1])
```

```python
print(l[:-1])
```

```python
print(l[-1][0:3])
```

**Append()**-To insert a single element, the element will be added in the end

```python
lis=[23,45,78,90,"hello",True]
lis.append(89)
print(lis)
```

```python
lis.append([1,2,3])
print(lis)
```

**extend**- To extend a list with the element of another list

```python
l1=[1,2,3,4,"hello"]
l2=[5,6,7,8,9]
t=(10,11,12,13)
l1.extend(l2)
print(l1)
```

```python
l1.extend(t)
print(l1)
```

**remove(element)**-To remove the purticular element or value

```python
l1=[1,2,3,4,"hello",34,2,2,1]
l1.remove("hello")
print(l1)
```

```python
l1.remove(2)
print(l1)
```

**insert()**- Inserting an element into the specified position

```python
list=[23,45,78,90,"hello",True]
list.insert(0,"hi")
print(list)
```

```python
list.insert(-1,2)          #here -1 position will be the second last one bcz append() is used to inserting value to the end
print(list)
```

```python
list.insert(-2,False)     
print(list)
```

```python
list.insert(-3,"th")
print(list)
```

**pop(position)**- Delete the value from the position

```python
l1=[1,2,3,4,"hello",34,2,2,1]
l1.pop(1)
print(l1)
```

```python
l1.pop(-1)    #remove the element in the last position
print(l1)
```

```python
l1.pop()   #remove the last one
print(l1)
```

**del keyword**- It can be used to delete the value from the position and also used to delete the whole list 

```python
l1=[1,2,3,4,"hello",34,2,2,1]
del l1[0]
print(l1)
```

```python
del l1[3]
print(l1)
```

```python
l1=[1,2,3,4]
del l1 
```
