# What is an Error?

An Error is any problem that stops the program from executing correctly.

````python
print("Hello"   # an error
      # '(' was never closed
      # The program won't even start.
````

# What is an Exception?

* An exception is an error that occurs during program execution (runtime), interrupting the normal flow of the program.

````python
x = 10 / 0                      # ZeroDivisionError
print("This line will not execute.")
````

The program started correctly but stopped during execution.

# Difference Between Errors and Exceptions

| Error                              | Exception                       |
| ---------------------------------- | ------------------------------- |
| Occurs before execution            | Occurs during execution         |
| Problems in code logic or syntax | The problem is not in the code logic
| Program won't start                | Program starts then stops       |
| Cannot be handled using try-except | Can be handled using try-except |
Example: `SyntaxError`, `IndentationError` | Example:  `ZeroDivisionError`, `ValueError`, `FileNotFoundError` |



# The _try_ and _except_ Block

* Used to catch and handle exceptions gracefully.

>Syntax of try-except

```
try:
    # Risky code

except ExceptionType:
    # Code if exception occurs
```



##### _try_ block:

* The _try_ block in Python is used to test code that might raise an error. It lets you run risky operations without crashing your program if something goes wrong.


- It runs the code inside it.

- If everything works fine, it continues normally.

- If an error occurs, it jumps to the matching except block to handle the error.

##### _except_ block:

* The except block in Python is used to catch and handle errors that occur inside a try block.


- It intercepts specific exceptions (errors) raised in the try block.

- It prevents the program from crashing.

- It lets you respond to errors with custom messages or recovery actions.


````python
age = int(input("Enter your age: "))
print(age)
````

````python
try:
    age = int(input("Enter your age: "))
    print(age)

except ValueError:
    print("Please enter numbers only.")
````

````python
try:
    x = int("hello")                  # This will raise a ValueError
except ValueError:
    print("Oops! That's not a number.")
````

````python
try:
    num1 = int(input("Enter first number: "))
    num2 = int(input("Enter second number: "))

    result = num1 / num2

    print("Result =", result)

except ZeroDivisionError:
    print("Cannot divide by zero.")
````

##### _else_ Block:

- The else block runs only if no exception occurs in the try block.

````python
try:
    x = int(input("Enter a number: "))
    y = 10 / x
except ZeroDivisionError:
    print("Cannot divide by zero.")
except ValueError:
    print("Please enter a valid integer.")
else:
    print("SUCCESSFULLY DONE")
````

````python
try:
    num = int(input("Enter number: "))
except ValueError:
    print("Invalid input.")
else:
    print("You entered:", num)
````

````python
print("Simple Calculator")

try:

    num1 = float(input("Enter first number: "))
    num2 = float(input("Enter second number: "))

    operation = input("Enter (+,-,*,/): ")

    if operation == "+":
        print(num1 + num2)

    elif operation == "-":
        print(num1 - num2)

    elif operation == "*":
        print(num1 * num2)

    elif operation == "/":
        print(num1 / num2)

    else:
        print("Invalid Operator")

except ValueError:
    print("Enter numbers only.")

except ZeroDivisionError:
    print("Cannot divide by zero.")
````

##### _finally_ Block:

* The finally block always executes, whether an exception occurs or not

````python
try:
    f = open("demo.txt", "r")
    print(f.read())
except FileNotFoundError:
    print("File not found.")
finally:
    print("Execution completed.")
````

### Multiple Exceptions"

````python
try:
    number = int(input("Enter number: "))

    result = 100 / number

    print(result)

except ValueError:
    print("Invalid Number")

except ZeroDivisionError:
    print("Cannot divide by zero")
````

* You can handle multiple exceptions using a tuple.

````python
try:
    x = int(input("Enter number: "))
    y = 10 / x
except (ZeroDivisionError, ValueError):
    print("Error: Invalid operation.")
````

### Catching All Exceptions (Generic Exception)

* Use Exception as a base class to catch any kind of runtime error.

````python
try:
    x = int("abc")
except Exception as e:
    print("Error occurred:", e)
````

````python
try:
    num = int(input("Enter Number: "))
    print(10 / num)

except Exception as e:
    print("Error:", e)
````

### Using raise to Manually Trigger Exceptions

* You can raise exceptions using the raise keyword.

````python
age = int(input("Enter your age: "))
if age < 0:
    raise ValueError("Age cannot be negative.")
else:
    print("Age is valid.")
````

````python
try:
    marks = int(input("Enter Marks: "))
    percentage = (marks / 500) * 100
    print("Percentage =", percentage)

except ValueError:
    print("Marks should be numbers.")
````

````python
students = ["John", "Alice", "Rahul"]

try:
    index = int(input("Enter Index: "))
    print(students[index])

except IndexError:
    print("Student not found.")
````

````python
student = {
    "name": "John",
    "age": 20
}

try:
    print(student["course"])

except KeyError:
    print("Key does not exist.")
````

### Custom (User-Defined) Exceptions

* You can define your own exceptions by creating a class that inherits from Exception.

````python
class InvalidAgeError(Exception):
    pass

try:
    age = int(input("Enter age: "))
    if age < 18:
        raise InvalidAgeError("You must be 18 or older.")
except InvalidAgeError as e:
    print(e)
````

````python
class AgeError(Exception):
    pass
try:
    age=10
    if age<18:
        raise AgeError("Not a valid age")
except Exception as e:
    print(e)
````
